# **Charts Design**

## Monthly sales trends

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['YearMonth'] = df['Order Date'].dt.to_period('M').astype(str)
monthly_revenue = df.groupby('YearMonth')['Price (INR)'].sum().reset_index()

plt.figure()
plt.plot(monthly_revenue['YearMonth'],monthly_revenue['Price (INR)'])
plt.xlabel('YearMonth')
plt.ylabel('Total Sales (INR)')
plt.title('Monthly Revenue Trends')
plt.xticks(rotation=45)
plt.tight_layout()
plt.grid(True)
plt.show()
```

## Daily Sales Trend

```python
df['DayName']= pd.to_datetime(df['Order Date']).dt.day_name()
daily_revenue = df.groupby('DayName')['Price (INR)'].sum().reindex(['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday'])

plt.figure(figsize=(10,5))
plt.bar(daily_revenue.index,daily_revenue.values)
plt.xlabel('Day of the Week')
plt.ylabel('Total Sales (INR)')
plt.title('Daily Revenue Trends(Mon-Tue)')
plt.xticks(rotation=45)
plt.show()
```

## Total Sales by Food Type (Veg& Non-Veg)

```python
non_veg_keywords = [
    "chicken","egg","fish","mutton","prawn","biryani","kabab","kebab","non-veg","non veg"
]
df["Food Category"] = np.where(
    df["Dish Name"].str.lower().str.contains('|'.join(non_veg_keywords),na = False),
    "Non-Veg","Veg")
```

```python
food_revenue = (
    df.groupby('Food Category')['Price (INR)']
    .sum().reset_index()
)
```

```python
fig = px.pie(
    food_revenue,
    values = 'Price (INR)',
    names = 'Food Category',
    hole = 0.5,
    title = 'Revenue Contribution by Food Category: Veg & Non-Veg'
)
fig.update_traces(textinfo='percent+label', pull=[0.05,0])

fig.update_layout(
    height = 500,
    margin = dict(t=60,b=40,l=40,r=40)
)
fig.show()
```

## Total Sales by States

```python
fig = px.bar(
    df.groupby('State',as_index=False)['Price (INR)'].sum().sort_values(by='Price (INR)',ascending=False),
    x = 'Price (INR)',
    y = 'State',
    orientation = 'h',
    color = 'Price (INR)',
    # color_continuous_scale = 'reds',
    title = 'Revenue by State(INR)',
)
fig.update_layout(height=600,yaxis=dict(autorange='reversed'))
fig.show()
```

## Quarterly Performance Summary

```python
df["Order_Date"] = pd.to_datetime(df["Order Date"])
df["Quarter"] = df["Order_Date"].dt.to_period("Q").astype(str)
quarterly_summary = (
    df.groupby("Quarter" ,as_index=False)
    .agg(
        Total_Sales=("Price (INR)", "sum"),
        Total_Orders=("Order_Date", "count"),
        Avg_Rating=("Rating", "mean"))
    .sort_values(by="Quarter")
    )


# quarterly_summary["Total_Sales"] = quarterly_summary["Total_Sales"].round(0)

# quarterly_summary["Avg_Rating"] = quarterly_summary["Avg_Rating"].round(2)

quarterly_summary
```

## Top 5 Cities By Sales

```python
top_5_cities = (
    df.groupby("City")["Price (INR)"]. sum()
    .nlargest(5)
    .sort_values().reset_index()
)

fig = px.bar(top_5_cities,
             x = "Price (INR)",
             y = "City",
             orientation = "h",
             color = "Price (INR)",
             title = "Top 5 Cities by Sales (INR)",
             color_discrete_sequence =['red']
)
# fig.update_layout(height = 600)
fig.show()
```

## Weekly trend Analysis

```python
df['Week'] = df['Order Date'].dt.isocalendar().week.astype(int)
weekly_revenue = df.groupby('Week')['Price (INR)'].sum().reset_index()

fig = px.line(
    weekly_revenue,
    x = 'Week',
    y = 'Price (INR)',
    title = 'Weekly Sales Trend'
)

fig.update_xaxes(dtick=1)
fig.update_layout(
    hovermode='x unified',
    height=500,
    margin=dict(t=60, b=40, l=40, r=40)
)
fig.show()
```
